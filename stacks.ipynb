# Stack
A stack is a linear data structure that operates on the Last In, First Out (LIFO) principle. Perfect for temporary storage and backtracking

## Last In, First Out (LIFO)
The lastest added element in is the first one to be removed out. A stack can be visualized as a vertical pile of plates, where you can only add or remove plates from the top. Elements are added and removed from the same end, called the "top" of the stack.

## Operations
+ push: Add an element to the top of the stack
+ pop: Remove the top element from the stack
+ peek, top: View the top element without removing it
+ is_empty: Check if the stack is empty
+ size: Get the number of elements

## Real-World Applications

+ Plate piles
+ Function calls: Your program's call stack
+ Undo operations: Ctrl+Z in text editors
+ Browser history: Back button functionality
+ Expression evaluation: Parentheses matching
+ Syntax parsing

## AbstractStack
The AbstractStack class serves as a contract that defines the essential interface for all stack implementations. It exists to ensure consistency across different stack implementations and to leverage Python's type system for better code reliability.
- Provides a standardized interface for all stack implementations
- Enforces implementation of core stack operations
- Makes code more maintainable and extensible
- Enables polymorphism. You can use different stack types interchangeably
- Facilitates unit testing and code maintainability
- Ensures consistent API across implementations

## ListStack
The simplest stack implementation that uses Python's built-in list data structure to provide stack functionality. It treats the end of the list as the top of the stack, leveraging the list's efficient append and pop operations.
- Leverages Python's optimized list operations
- Automatic memory management and resizing
- Good cache locality due to contiguous memory layout
- Built-in bounds checking and error handling

```python
from typing import Any


class AbstractStack(ABC):
    """Abstract base class for LIFO stack implementations."""

    @abstractmethod
    def push(self, item: Any) -> None:
        """Add an item to the top of the stack."""
        pass

    @abstractmethod
    def pop(self) -> Any:
        """Remove and return the top item from the stack.

        Raises:
            StackEmpty: If the stack is empty.
        """
        pass

    @abstractmethod
    def peek(self) -> Any:
        """Return the top item without removing it.

        Raises:
            StackEmpty: If the stack is empty.
        """
        pass

    @abstractmethod
    def is_empty(self) -> bool:
        """Return True if the stack is empty, False otherwise."""
        pass

    @abstractmethod
    def size(self) -> int:
        """Return the number of items in the stack."""
        pass

    def __len__(self) -> int:
        """Return the number of items in the stack."""
        return self.size()

    def __bool__(self) -> bool:
        """Return True if the stack is not empty."""
        return not self.is_empty()


class StackEmpty(Exception):
    """Exception raised when trying to pop from an empty stack."""
    pass
```

```python
class ListStack(AbstractStack):
    """LIFO stack implementation using a Python list."""

    def __init__(self):
        """Initialize an empty stack."""
        self._items = []

    def push(self, item: Any) -> None:
        self._items.append(item)

    def pop(self) -> Any:
        if self.is_empty():
            raise StackEmpty("Cannot pop from empty stack")
        return self._items.pop()

    def peek(self) -> Any:
        if self.is_empty():
            raise StackEmpty("Cannot peek at empty stack")
        return self._items[-1]

    def is_empty(self) -> bool:
        return len(self._items) == 0

    def size(self) -> int:
        return len(self._items)

    def __str__(self) -> str:
        """Return string representation of the stack."""
        return f"Stack({self._items})"

    def __repr__(self) -> str:
        """Return detailed string representation."""
        return f"Stack({self._items})"
```

```python
# Example usage and demonstration
if __name__ == "__main__":
    print("=== ListStack Demo ===")
    stack1 = ListStack()

    # Test basic operations
    print(f"Empty stack: {stack1}")
    print(f"Is empty: {stack1.is_empty()}")

    # Push items
    for i in [1, 2, 3, 4, 5]:
        stack1.push(i)
        print(f"Pushed {i}: {stack1}")

    print(f"Size: {stack1.size()}")
    print(f"Peek: {stack1.peek()}")

    # Pop items
    while not stack1.is_empty():
        item = stack1.pop()
        print(f"Popped {item}: {stack1}")

    print("\n=== LinkedStack Demo ===")
    stack2 = ListStack()

    # Test with strings
    words = ["hello", "world", "python", "stack"]
    for word in words:
        stack2.push(word)
        print(f"Pushed '{word}': {stack2}")

    print(f"Size: {len(stack2)}")  # Using __len__

    # Pop all items
    while stack2:  # Using __bool__
        item = stack2.pop()
        print(f"Popped '{item}': {stack2}")

    print("\n=== Error Handling Demo ===")
    empty_stack = ListStack()

    try:
        empty_stack.pop()
    except StackEmpty as e:
        print(f"Caught exception: {e}")

    try:
        empty_stack.peek()
    except StackEmpty as e:
        print(f"Caught exception: {e}")

    print("\n=== Practical Example: Balanced Parentheses ===")


    def is_balanced(expression: str) -> bool:
        """Check if parentheses in expression are balanced."""
        stack = ListStack()
        pairs = {'(': ')', '[': ']', '{': '}'}

        for char in expression:
            if char in pairs:  # Opening bracket
                stack.push(char)
            elif char in pairs.values():  # Closing bracket
                if stack.is_empty():
                    return False
                if pairs[stack.pop()] != char:
                    return False

        return stack.is_empty()
```

# Queues
A queue is a linear data structure that operates on the First In, First Out (FIFO) principle. Ideal for scheduling and processing in order
## First In, First Out (FIFO)
The first element added in to the queue is the first one to be removed out. A queue can be visualized as a line of people waiting on a line at a grocery store, where people are served in the order they arrive.
### Operations
+ Enqueue: Add an element to the rear of the queue
	+ Time Complexity:  O(1) for basic operations
    + Space Complexity: O(n) where n is the number of elements
+ Dequeue: Remove an element from the front of the queue
+ Front: View the front element without removing itg it
+ isEmpty: Check if the queue is empty

### Real-World Applications
+ Customer service: Call center wait times
+ Print jobs: Documents waiting to be printed
+ Task scheduling: Operating system process queues

```python
class BaseQueue(ABC):
    """
    Queue implementation using Python deque for optimal performance
    Follows FIFO (First In, First Out) principle
    """

    def __init__(self, capacity=None):
        """Initialize queue with optional capacity"""
        self._size = 0
        self._capacity = capacity

    @abstractmethod
    def get_items(self):
        """Get items for displays"""
        pass

    @abstractmethod
    def enqueue(self, item):
        """Add item to rear of queues"""
        pass

    @abstractmethod
    def dequeue(self):
        """Remove and return front items"""
        pass

    @abstractmethod
    def front(self):
        """View front item without removings"""
        pass

    # Common methods that all queues share
    def is_empty(self):
        return self._size == 0

    def size(self):
        return self._size

    def is_full(self):
        if self._capacity is None:
            return False
        return self._size >= self._capacity

    def display(self):
        if self.is_empty():
            print(f"{self.__class__.__name__}: [] (empty)")
            return
        items = self.get_items()
        print(f"{self.__class__.__name__}: {items}")

    def __str__(self):
        if self.is_empty():
            return f"{self.__class__.__name__}([])"
        items = self.get_items()
        return f"{self.__class__.__name__}({items})"

    def __len__(self):
        return self._size


class Queue(BaseQueue):
    """ Trivial Queue """

    def __init__(self):
        super().__init__()
        self.items = []

    def enqueue(self, item):
        """Add an item to the rear of the queue"""
        self.items.append(item)
        print(f"Enqueued {item} into queue")

    def dequeue(self):
        """Remove and return the front item from the queue"""
        if self.is_empty():
            raise IndexError("Queue is empty")
        item = self.items.pop(0)
        print(f"Dequeued {item} from queue")
        return item

    def front(self):
        """Return the front item without removing it"""
        if self.is_empty():
            raise IndexError("Queue is empty")
        print("viewing the front item")
        return self.items[0]

    def get_items(self):
        """Display the queue contents (front to rear)"""
        return self.items
```
